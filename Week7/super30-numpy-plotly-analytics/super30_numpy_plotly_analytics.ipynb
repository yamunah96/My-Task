# **Data Generation**


```python
import numpy as np
np.random.seed(42)

months=np.array(["jan","feb","march","april","may","june","july","aug","sep","oct,","nov","dec"])
print("Months: ",months)

# im considering revenue of a company from 2Lacs to 5Lacs range
revenue=np.random.randint(200000,500000,12)
print("Revenue: ",revenue)

# orders - number of orders placed in each month-  lets take 500-3000
orders= np.random.randint(500,3000,12)
print("Orders: ",orders)

# customers -number of customers placed order in each month lets take 100-2000
customers= np.random.randint(100,2000,12)
print("Customers: ",customers)
# expenses  the total amount company spedning in that month
expenses= np.random.randint(40000, 120001, 12)
print("Expenses: ",expenses)
```

#**NumPy Processing and Analysis**

```python
# Calculate total annual revenue.
print("Total annual revenue: ",np.sum(revenue))

# Calculate average monthly revenue.
print("Average monthly revenue: ",np.mean(revenue))

#Find the best-performing month.
best_performing_month= months[np.where([i >= np.max(revenue) for i in revenue])]
print("best_performing_month: ",best_performing_month)

# Find the lowest-performing month.
lowest_performing_month= months[np.where([i <= np.min(revenue) for i in revenue])]
print("lowest_performing_month: ",lowest_performing_month)

# Calculate month-to-month revenue differences.
revenue_differences=np.diff(revenue)
revenue_differences
```

#**Plotly Visualization**

```python
  # Create a line chart showing revenue trends.
import plotly.express as px
revenue_trends=px.line(x=months,y=revenue,labels={"x":"Months","y":"Revenue"},title="Monthly Revenue Trends",markers=True,text=revenue)

revenue_trends.update_traces(
    line=dict(color="green"),
    marker=dict(color="red")
)

revenue_trends.show()
```

```python
# Create a bar chart showing monthly orders.
import plotly.express as px
bar_graph= px.bar(x=months,y=orders,title="Monthly Orders",labels={"x": "Month","y": "Orders"},orientation="v")

bar_graph.show()
```

```python
# Create a comparison graph for Revenue, Expenses
import plotly.express as px
bar_graph= px.bar(x=months,y=[revenue,expenses],title="Monthly Revenue vs Expenses",
  labels={
        "x": "Month",
        "y": "Amount"
    },
    orientation="v",
    barmode="group",
)

bar_graph.show()
```

```python
import numpy as np
import plotly.graph_objects as go

fig = go.Figure()

# Revenue bars
fig.add_trace(
    go.Bar(
        x=months,
        y=revenue,
        name="Revenue",
        text=revenue,
        textposition="outside",
        hovertemplate="Month: %{x}<br>Revenue: ₹%{y}<extra></extra>"
    )
)

# Expenses bars
fig.add_trace(
    go.Bar(
        x=months,
        y=expenses,
        name="Expenses",
        text=expenses,
        textposition="outside",
        hovertemplate="Month: %{x}<br>Expenses: ₹%{y}<extra></extra>"
    )
)

fig.update_layout(
    title="Monthly Revenue vs Expenses",
    xaxis_title="Month",
    yaxis_title="Amount (₹)",
    barmode="group"
)

fig.show()
```

```python
# Create customer-growth visualization.
import plotly.express as px

customer_growth = px.line(
    x=months,
    y=customers,
    title="Monthly Customer Growth",
    labels={
        "x": "Month",
        "y": "Number of Customers"
    },
    markers=True,
    text=customers
)

customer_growth.update_traces(
    textposition="top center",
    hovertemplate="Month: %{x}<br>Customers: %{y}<extra></extra>"
)

customer_growth.show()
```

```python
# Calculate profit Profit = Revenue - Expenses and visualize monthly profit.
profit= revenue-expenses
bar_graph= px.bar(x=months,y=profit,title="Monthly Profit",labels={"x": "Month","y": "Profit"},orientation="v")
bar_graph.update_traces(
    textposition="outside",
    hovertemplate="Month: %{x}<br>Profit: ₹%{y:,.0f}<extra></extra>"
)

bar_graph.show()
```
